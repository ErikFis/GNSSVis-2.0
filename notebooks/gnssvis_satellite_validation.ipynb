# GNSSVis Validation - GNSS Satellite Identification

Use the unified API to inspect satellite identification in UIRAPURU radio
telescope observations. Set FITS_FILE to a local observation before running.


```python
import sys, os
import warnings
warnings.filterwarnings('ignore')

modules_to_remove = [mod for mod in sys.modules.keys() if 'gnssvis' in mod]
for mod in modules_to_remove:
    del sys.modules[mod]

from gnssvis.core import GNSSVis

FITS_FILE = '../data/Horn.fits'
gv = GNSSVis()
gv.plot_waterfall_with_sky(FITS_FILE, beamwidth_deg=12.0)
```

```python
subtracted = gv.baseline_subtract()
```

```python
smoothed = gv.gaussian_smooth(subtracted, sigma_freq=10.0, sigma_time=10.0, plot=True)
```
